```python
# Name: Alexander Ehrich
# Date: Jun 22, 2022


# we import the libraries
import json
from pprint import pprint


movies = []
with open("movies.json", encoding="utf8") as f:  # we open the file, specifyng the encoding
    for line in f:
        movies.append(json.loads(line))          # We load the file

        
        
# 1. How many movies are included in the file?
def total_movies(data_path):
    count = 0
    movies = []
    
    for element in data_path:    # we iterate inside of the file
        for title in element:
            movies.append(title["title"]) # we append all the titles to an empty list
    
    return (f"There are a total of {len(movies)} movies in this dataset") # we print the lenght of this list



# 2. In what year was the movie A Separation made?
def year_movie(movie_title, data_path):
    year = ""
    
    for element in data_path:
        for title in element:
            if title["title"] == movie_title:   # if a title form the file is the same as the input title,
                year = title["year"]            # the year of that movie is stored in a variable
    
    return (f"{movie_title} was made in {year}")  # the movie title and year are printed



# 3. How many movies has Martin Scorsese directed?
def director_movies(director_name, data_path):
    count = 0
    
    for element in data_path:
        for director in element:
            if director["directors"] == [director_name]:   # if a director's name from the file matches to the input name,
                count += 1                                 # a counter is increased
            else:
                count = None
                
    return (f"{director_name} has directed {count} movies")  # it returns the total amount of movies



# 4. Which is the shortest movie of the dataset?
def shortest_movie(data_path):
    duration = 1000
    title = []
    
    for element in data_path:
        for dur in element:
            if int(dur["runtime"]) < duration:       # if the runtime of a movie is shorter than the starting one,
                duration = int(dur["runtime"])       # that movie will be the new shortest one
                title.append(dur["title"])
                
            elif int(dur["runtime"]) == duration:    # if there are more than one with the same duration, theyr name
                title.append(dur["title"])           # is added to a title's list, to be printed at the end
    
    if len(title) == 1:
        return ("The shortest movie of this dataset is", print(*title, sep=", "),f"with a runtime of {duration} minutes")
    
    else:
        print(*title, sep=", ") 
        print(f"Are the shortest movies with a runtime of {duration} minutes")


        
# 5. Which is the longest movie of the dataset?
def longest_movie(data_path):
    duration = 0
    title = []
    
    for element in data_path:
        for dur in element:
            if int(dur["runtime"]) > duration:       # if the runtime of a movie is longer than the starting one,
                duration = int(dur["runtime"])       # that movie will be the new longest one
                duration = int(dur["runtime"])
                title.append(dur["title"])
                
            elif int(dur["runtime"]) == duration:    # if there are more than one with the same duration, theyr name
                title.append(dur["title"])           # is added to a title's list, to be printed at the end
                title.append(dur["title"])
    
    if len(title) == 1:
        return ("The longest movie of this dataset is", print(*title, sep=", "),f"with a runtime of {duration} minutes")
    
    else:
        print(*title, sep=", ") 
        print(f"Are the longest movies with a runtime of {duration} minutes")


            
#Section to view the result, works with inputs, the user is told what to do at any time
print("[1] If you want to know the total amount of movies of the dataset")
print("[2] If you want to know the year a movie was made")
print("[3] If you want to know how many movies a director has directed")
print("[4] If you want to know the shortest movies of the dataset")
print("[5] If you want to know the longest movies of the dataset")

input_1 = input("Hello, what do you want to know? [add the number of the information you want to know]")
input_1 = int(input_1)


if input_1 == 1:
    print(total_movies(movies))

elif input_1 == 2:
    input_2 = input("Please type the title of the movie you want to know the year of: ")
    print(year_movie(input_2, movies))

elif input_1 == 3:
    input_3 = input("Please type the name of the director from who you want to know the amount movies:")
    print(director_movies(input_3, movies))

elif input_1 == 4:
    print(shortest_movie(movies))

elif input_1 == 5:
    print(longest_movie(movies))

else:
    print("Please insert a valid number.")
```
