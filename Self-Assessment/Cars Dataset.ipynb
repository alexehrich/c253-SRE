```python
# Name: Alexander Ehrich
# Date: Jun 26, 2022


# 1. Convert file into a list where each element is a dictionary (key = columns names & values = elements of each column)
#reads CSV file and returns list
def fromCSV(path,delimiter,quotechar):
    import csv
    from pprint import pprint
    
    # import the csv module
    data = list()                           # any data we will read will be returned in a list 
    with open(path, newline='') as csvfile: # open the file
        filecontent = csv.DictReader(csvfile, delimiter=delimiter, quotechar=quotechar) 
        # access the content of the file
        for row in filecontent:             # iterate through the rows 
            data.append(row)                # save the rows in the data list. Each row is a dictionary 
    #pprint(data)
    return data

     
               
# 2. Indentify records with a specific 'make' input by the user
def print_make():
    makes = []
    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    
    for element in data:
        if element["make"] not in makes:
            makes.append(element["make"])
        else:
            continue
    
    print("Which of the following makes do you want to check?")
    for i in makes:
        print(i)
    
    while True:
        input_make = input(" Please insert a car make: ")
        input_make = input_make.lower()
        if input_make in makes:
            break
        elif input_make == "stop":
            break
        else:
            print("Please, write a valid name")
    
    data_filtered = filter(lambda x: x['make'] == input_make, data) 
    data_filtered_list = list(data_filtered) #convert the filter object into a list
    for row in data_filtered_list: #display each element in the filtered list 
        print(f"{row}\n")
    return



# 3. Identify all fuel-efficient cars
def fuel_efficient():
    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    
    print("The following cars are considered as fuel-efficient vehicles: \n")
        
    data_filtered = filter(lambda x: int(x['city_mpg']) > 35, data) 
    data_filtered_list = list(data_filtered) #convert the filter object into a list
    for row in data_filtered_list: #display each element in the filtered list 
        print(f"{row}\n")
    return



# 4. Identify cars with horsepower greater than 100
def high_horsepower():
    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    data_new = []
    for element in data:
        if element["horsepower"] != "?":
            data_new.append(element)
        else:
            continue
    
    print("The following cars have more than 100 hp: \n")
        
    data_filtered = filter(lambda x: float(x['horsepower']) > 100, data_new) 
    data_filtered_list = list(data_filtered) #convert the filter object into a list
    for row in data_filtered_list: #display each element in the filtered list 
        print(f"{row}\n")
    return



# 5. Calculate the cost of driving 100 miles ib tge city, after user has input a gas cost per gallon
def cost_of_driving():
    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    diesel_cars = []
    gas_cars = []
    
    for element in data:
        if element["fuel_type"] == "gas":
            gas_cars.append(element)
        else:
            diesel_cars.append(element)
    
    print("These are the costs of driving 100 miles in a city: \n")
    
    while True:
        input_fuel = input(" Does your car use gas or diesel? \n")
        input_fuel = input_fuel.lower()
        
        if input_fuel == "diesel":
            input_3 = input("How much does a diesel gallon cost? \n")
            input_3 = float(input_3)
            
            map_iterator = map(lambda x: input_3*(100/float(x["city_mpg"])), diesel_cars)
            map_list = list(map_iterator)
            
            for i in range(0,len(diesel_cars)):
                diesel_cars[i]['cost_100_city_miles'] = round(map_list[i],2)
                print(f"{diesel_cars[i]} has a cost for 100 city miles = {diesel_cars[i]['cost_100_city_miles']} $ \n")
            break
            
        elif input_fuel == "gas":
            input_3 = input("How much does a gas gallon cost ($)? \n")
            input_3 = float(input_3)
            
            map_iterator = map(lambda x: input_3*(100/float(x["city_mpg"])), gas_cars)
            map_list = list(map_iterator)
            
            for i in range(0,len(gas_cars)):
                gas_cars[i]['cost_100_city_miles'] = round(map_list[i],2)
                print(f"{gas_cars[i]} has a cost for 100 city miles = {gas_cars[i]['cost_100_city_miles']} $ \n")
            break
        
        elif input_fuel == "stop":
            break
        
        else:
            print("Sorry, we only offer diesel or gas cars. \n")
            break
    return



# 6. Average city and highway MPG across all cars.
def avg_mpg():
    from functools import reduce

    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    city_mpg = []
    highway_mpg = []
    
    for element in data:
        city_mpg.append(int(element["city_mpg"]))
        highway_mpg.append(int(element["highway_mpg"]))

    result_city = reduce(lambda x,y: x+y, city_mpg)
    result_city = result_city/len(city_mpg)
    print(f"The average city consume of the cars we offer is: {round(result_city,2)} (miles/gallon)")
    
    result_highway = reduce(lambda x,y: x+y, highway_mpg)
    result_highway = result_highway/len(highway_mpg)
    print(f"The average highway consume of the cars we offer is: {round(result_highway,2)} (miles/gallon)")
    return



# 7. Average cost of all cars.
def avg_cost():
    from functools import reduce
    
    data = fromCSV(path='../Data/car_data.csv',delimiter=',',quotechar='|') 
    costs = []
    
    for element in data:
        if element["price"] != "?":
            costs.append(int(element["price"]))
        else:
            continue

    result = reduce(lambda x,y: x+y, costs)
    result = result/len(costs)
    print(f"The average cost of the cars we offer is: {round(result,2)} $")
    return



if __name__ == "__main__":
    print("1. Convert file into a list\n2. Indentify records with a specific 'make' input by the user\n3. Identify all fuel-efficient cars\n4. Identify cars with horsepower greater than 100\n5. Calculate the cost of driving 100 miles ib tge city, after user has input a gas cost per gallon\n6. Average city and highway MPG across all cars\n7. Average cost of all cars.")
    
    review = input("Please insert a number from 1 to 7 depending on what you want to check: \n")
    
    if review == "1":
        fromCSV(path,delimiter,quotechar)
    elif review == "2":
        print_make()
    elif review == "3":
        fuel_efficient()
    elif review == "4":
        high_horsepower()
    elif review == "5":
        cost_of_driving()
    elif review == "6":
        avg_mpg()
    elif review == "7":
        avg_cost()
        next
    elif review == "stop":
        print("Bye!")
    else:
        print("Please write an int or stop")
```
