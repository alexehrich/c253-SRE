```python
# Name: Alexander Ehrich
# Date: Jun 20, 2022


# imports libraries
from random import random 
from random import seed 
import string
import secrets

# we createg a string, which contains all letters of the alphabet, numbers & special characters
password_characters = string.ascii_letters + string.digits + "!#$%&'()*+,-./:;<=>?@[\]^_`{|}~"
print(password_characters)

class PasswordGenerator: # we define the class
    def __iter__(self): # iteration function
        return self  
    def __next__(self):
        # a blank string is created to later add the password content
        password = "" 
        # a for loop is added to iterate through the process 15 times
        for i in range(0,15): 
            random_number = secrets.choice(password_characters) # generates a random value
            password = password + str(random_number) # the generated random value is added to the password string
        return password

result = [] # a blank list is created to add the five generated passwords

# the class and the iteration are defined into a variable for a leter execution
passwords = PasswordGenerator()
passwords_iterator = iter(passwords) 

result.append(next(passwords_iterator)) # the generated passwords are appended to the list
result.append(next(passwords_iterator))
result.append(next(passwords_iterator))
result.append(next(passwords_iterator))
result.append(next(passwords_iterator))

# the list with the 5 passwords is printed
print(result)

# the user has to select which password he/she likes more (or if none of them is apropiate)
value = int(input("Please, select the password that suits best to you [1 (first password) to 5 (last password), 0 (if none of the passwords is suitable)]: "))

# depending on the selection of the user, it will show one or another password
if value == 1:
    print('Your chosen password has been: ', result[0])
if value == 2:
    print('Your chosen password has been: ', result[1])
if value == 3:
    print('Your chosen password has been: ', result[2])
if value == 4:
    print('Your chosen password has been: ', result[3])
if value == 5:
    print('Your chosen password has been: ', result[4])
if value == 0:
    print('Please try again and we hope you find a more suitable password :)')
```
