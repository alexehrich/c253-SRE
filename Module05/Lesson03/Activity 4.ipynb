```python
# Download this file to use for this activity: stocks.zip.

    # You will need to extract or uncompress the zip file before using the data.
    
    # The zipped file contains many text files, each of which represents the stocks for one particular company. 
    # You should treat this collection as a single dataset rather than as individual files when solving the problems 
    # listed below.

# Write a program to retrieve data to find the following information:

# 1. Return a list of all company indexes that appear at least once in the dataset, sorted alphabetically by index value.

# 2. Return the number of publicly traded companies for each year in the dataset.

# 3. Return a list of indexes of companies that were publicly traded in each year in the dataset, grouped by year.
    # For example, the output should look something like this:
        # [ "Year 1":[index1,index2,index3,….],
        # "Year 2": [index1,index2,index3,….],
        # …]
        
# 4. Choose a company index value and return all historical pricing data for the company.

# 5. Choose a company index value and year and return the pricing data for the company for that particular year.

# 6. Choose a company index value, a year, and a month, and return all pricing data for the company for that particular 
#    year and month.

# 7. Choose a year and return the list of companies that were publicly traded during that year. The results should include
#    all companies that traded for at least one day during the given year.
```

```python
# Name: Alexander Ehrich
# Date: Jun 26, 2022


# 1. Return a list of all company indexes that appear at least once in the dataset, sorted alphabetically by index value.
def companies_indexes():
    import os
    dir_path = r'../../Data/stocks'   # folder path
    count = 0

    for path in os.listdir(dir_path):   # Iterate directory
        if os.path.isfile(os.path.join(dir_path, path)):   # check if current path is a file
            count += 1
            print(path)
    print(f'There is a total amount of {count} company indexes')
    
    return



# 2. Return the number of publicly traded companies for each year in the dataset.
def publicly_traded_companies_length():
    import os
    import re

    dir_path = r'../../Data/stocks'
    publiclyTraded = {}
    
    for path in os.listdir(dir_path):   # Iterate directory
        if os.path.isfile(os.path.join(dir_path, path)):   # check if current path is a file
            file = open("../../Data/stocks/"+path,'r')
            text = file.readlines()[1:]
            years = set(map(lambda s : s.split('-')[0],text))
            for year in years:
                if year in publiclyTraded.keys():
                    publiclyTraded[year] += 1
                else:
                    publiclyTraded[year] = 1
    
    print(publiclyTraded)
    return 
            

    
# 3. Return a list of indexes of companies that were publicly traded in each year in the dataset, grouped by year.
def publicly_traded_companies_name():
    import os
    import re
    from pprint import pprint

    dir_path = r'../../Data/stocks'
    publiclyTraded = {}
    
    for path in os.listdir(dir_path):   # Iterate directory
        if os.path.isfile(os.path.join(dir_path, path)):   # check if current path is a file
            file = open("../../Data/stocks/"+path,'r')
            text = file.readlines()[1:]
            years = set(map(lambda s : s.split('-')[0],text))
            for year in years:
                if year in publiclyTraded.keys():
                    publiclyTraded[year].append(path)
                else:
                    publiclyTraded[year] = [path]
                    
    pprint(publiclyTraded)
    return 



# 4. Choose a company index value and return all historical pricing data for the company.
def pricing_data(path):
    from pprint import pprint
    
    file = open("../../Data/stocks/"+path,'r')
    text = file.readlines()
    pprint(text)
    
    return
    

    
# 5. Choose a company index value and year and return the pricing data for the company for that particular year.
def pricing_data_year(path,year_input):
    from pprint import pprint
    import re
    content = []
    
    infile = open("../../Data/stocks/"+path,'r')
    firstLine = infile.readline()
    print(f"\n{firstLine}")
    infile.close()
    
    file = open("../../Data/stocks/"+path,'r')
    text = file.readlines()
    print(file.readline())
    content = list(map(lambda l: l.rstrip('\n'),filter(lambda x : x.split('-')[0] == str(year_input), text)))
    pprint(content)
    
    return



# 6. Choose a company index value, a year, and a month, and return all pricing data for the company for that particular 
#    year and month.
def pricing_data_year_month(path,year_input, mont_input):
    from pprint import pprint
    import re
    content = []
    date = year_input + "-" + mont_input
 
    infile = open("../../Data/stocks/"+path,'r')
    firstLine = infile.readline()
    print(f"\n{firstLine}")
    infile.close()
    
    file = open("../../Data/stocks/"+path,'r')
    text = file.readlines()
    print(file.readline())
    content = list(map(lambda l: l.rstrip('\n'),filter(lambda x : x.rsplit('-',1)[0] == str(date), text)))
    pprint(content)
    
    return



# 7. Choose a year and return the list of companies that were publicly traded during that year. The results should include
#    all companies that traded for at least one day during the given year.
def traded_companies_year(year_input):
    from pprint import pprint
    import re
    import os
    publiclyTraded = []
    dir_path = "../../Data/stocks/"
    
    for path in os.listdir(dir_path):   # Iterate directory
        if os.path.isfile(os.path.join(dir_path, path)):   # check if current path is a file
            file = open("../../Data/stocks/"+path,'r')
            text = file.readlines()[1:]
            years = set(map(lambda s : s.split('-')[0],text))
            for year in years:
                if year == year_input:
                    publiclyTraded.append(path)
                else:
                    continue
    
    print(publiclyTraded)
    
    return 



if __name__ == "__main__":
    while True:
        print("1. Return a list of all company indexes that appear at least once\n2. Return the number of publicly traded companies for each year\n3. List of indexes of companies that were publicly traded in each year\n4. Full pricing data for an input company\n5. Pricing data for the company for a particular year\n6. Pricing data for the company for a particular year and month\n7. List of companies that were publicly traded during an input year")

        review = input("Please insert a number from 1 to 7 depending on what you want to check: \n")

        if review == "1":
            companies_indexes()
        elif review == "2":
            publicly_traded_companies_length()
        elif review == "3":
            publicly_traded_companies_name()
        elif review == "4":
            path = input("Please input the file of the company index you want to check: ")
            pricing_data(path)
        elif review == "5":
            path = input("Please input the file of the company index you want to check: ")
            year_input = input("Please input a year: ")
            pricing_data_year(path, year_input)
        elif review == "6":
            path = input("Please input the file of the company index you want to check: ")
            year_input = input("Please input a year: ")
            month_input = input("Please input a month: ")
            pricing_data_year_month(path, year_input, month_input)
        elif review == "7":
            year_input = input("Please input a year: ")
            traded_companies_year(year_input)
            next
        elif review.lower() == "stop":
            print("Bye!")
            break
        else:
            print("Please write an int or stop")
```
