```python
# Name: Alexander Ehrich
# Date: Jun 28, 2022


import csv
from pprint import pprint
from functools import reduce


def read_csv(filepath,delimiter=","):
    dataset = list()
    try:
        with open(filepath) as f: 
            data = csv.DictReader(f, delimiter=delimiter) 
            for row in data:
                dataset.append(row)
    
    except FileNotFoundError:
        print("FileNotFoundError. Check the path or the filename again.)
 
    return dataset


              
# 1. Shows the artist with most songs in the Pop USA playlist.
def most_songs():
    data = read_csv("../../Data/EvolutionPopUSA_MainData.csv")
    
    artist_dict = {}
    artist_set = set(map(lambda a: a["artist_name_clean"], data))
    
    for artist in artist_set:
        artist_dict[artist] = len(list(filter(lambda art: art["artist_name_clean"] == artist, data)))
    
    most_songs = reduce(lambda x,y: x if artist_dict[x] > artist_dict[y] else y, artist_dict)
    
    print(f'{most_songs} is the artist with most songs, with a total of {artist_dict[most_songs]} songs')

    return
    

    
# 2. Identify the cluster with the most songs.
def cluster_most_songs():
    data = read_csv("../../Data/EvolutionPopUSA_MainData.csv")
    
    cluster_dict = {}
    cluster_set = set(map(lambda a: a["cluster"], data))
    
    for cluster in cluster_set:
        cluster_dict[cluster] = len(list(filter(lambda clust: clust["cluster"] == cluster, data)))
    
    cluster_most_songs = reduce(lambda x,y: x if cluster_dict[x] > cluster_dict[y] else y, cluster_dict)
    
    print(f'The cluster with most songs is number {cluster_most_songs}, with a total of {cluster_dict[cluster_most_songs]} songs')

    return



# 4. Identify the era with the most songs.
def era_most_songs():
    data = read_csv("../../Data/EvolutionPopUSA_MainData.csv")
    
    era_dict = {}
    era_set = set(map(lambda a: a["era"], data))
    
    for era in era_set:
        era_dict[era] = len(list(filter(lambda e: e["era"] == e, data)))
    
    era_most_songs = reduce(lambda x,y: x if era_dict[x] > era_dict[y] else y, era_dict)
    
    print(f'The era with most songs is the {era_most_songs}, with a total of {era_dict[era_most_songs]} songs')

    return
    


if __name__ == "__main__":
    most_songs()
    print("\n")
    cluster_most_songs()
    print("\n")
    era_most_songs()
```
