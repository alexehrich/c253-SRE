## Activity 2
Write a function that takes as input the path to two text files and concatenates the files into a single new file.

```python
def concat_files(path1, path2):
    text1 = ""
    text2 = ""
    
    f1 = open(path1, "r")
    text1 = text1 + f1.read()
    f1.close()
    
    f2 = open(path2, "r")
    text2 = text2 + f2.read()
    f2.close()
    
    input_1 = input("Please write the name of the wished new file: [Do not forget adding .txt] ")
    path3 = "../../Data/" + input_1
    
    f3 = open(path3, "w")
    f3.write(text1)
    f3.write("\n")
    f3.write(text2)
    f3 = open(path3, "r")
    print(f3.read())
    f3.close()

    
    
concat_files("../../Data/flatland.txt", "../../Data/randomtest.txt")
```

## Activity 3
Write a function that takes a variable number of text files and concatenates them into a single new file.

```python
def concat_n_files(*argv):
    
    input_1 = input("Please write the name of the wished new file: [Do not forget adding .txt] ")
    final_path = "../../Data/" + input_1 
    
    for arg in argv:
        text = ""
        f = open(arg, "r")
        text = text + f.read()
        f.close()
        
        final = open(final_path, "a")
        final.write(text)
        final.write("\n")
        text = ""   

    final = open(final_path, "r")
    print(final.read())
    final.close()

    
    
concat_n_files("../../Data/flatland.txt", "../../Data/randomtest.txt")
```

## Activity 4
Write a function that reads a text file line by line, stores each line in a list, and then returns the list.

```python
def read_and_store(path):
    lst = []
    line = []
    
    file = open(path, "r")   
    f = open(path, "r")
    
    for i in range(len(file.readlines())):
        line.append(f.readline())
        lst.append(line)
        line = []
        
    return lst

    

read_and_store("../../Data/flatland01.txt")
```

## Activity 5
Write a function that takes as input the path to a text file and returns the longest word in the text file.

```python
def longest_word(path):
    file = open(path,"r")
    read = file.read()
    words = read.split()
    max_len = len(max(words, key=len))
    for word in words:
        if len(word) == max_len:
            longest_word = word
    
    print(longest_word)
    
    
    
    
longest_word("../../Data/flatland01.txt")
```

## Activity 10: New York Restaurants
For this activity, use the file [restaurant.json](https://the-software-guild.s3.amazonaws.com/techstart-1909/data-files/restaurant.json); download it if you do not already have a copy.

This dataset includes records of restaurants in New York City, including where each restaurant is located, what kind of cuisine it serves, and its customer ratings. Here is a sample record:


    {"address": {"building": "1007", "coord": [-73.856077, 40.848447], "street": "Morris Park Ave", "zipcode": "10462"}, "borough": "Bronx", "cuisine": "Bakery", "grades": [{"date": {"$date": 1393804800000}, "grade": "A", "score": 2}, {"date": {"$date": 1378857600000}, "grade": "A", "score": 6}, {"date": {"$date": 1358985600000}, "grade": "A", "score": 10}, {"date": {"$date": 1322006400000}, "grade": "A", "score": 9}, {"date": {"$date": 1299715200000}, "grade": "B", "score": 14}], "name": "Morris Park Bake Shop", "restaurant_id": "30075445"}

Use Python to analyze the data in this file and find the following information:
1. Compute the average score for each restaurant.
1. Compute the minimum score for each restaurant.
1. Compute the maximum score for each restaurant.
1. Compute the average score for each type of cuisine in each borough.
1. Compute the minimum score for each type of cuisine in each borough.
1. Compute the maximum score for each type of cuisine in each borough.

```python
# Name: Alexander Ehrich
# Date: Jun 22, 2022

import json
from pprint import pprint

restaurant = []
with open("../../Data/restaurants.json") as file:
    for line in file:
        restaurant.append(json.loads(line))
             
# 1. Compute the average score for each restaurant.
def avg_score(data_path):
    avg_result = dict()
    avg_lst = []

    for element in data_path:    # iterates through: 'address', 'borough', cuisine, 'grades', etc.
        for score in element["grades"]:   # iterates through: 'date', 'grade', 'score' 
            if score["score"] != None:    # if the key score is not empty (has a value)
                avg_lst.append(score["score"])  # adds that value to a list, which stores all score values
        
        if len(avg_lst) != 0:             # if the are scores registered, then it calculates avg and appends to restaurant
            average = round(sum(avg_lst)/len(avg_lst), 2)
        else:                             # if there are not scores registered, it appends a None to the restaurant
            average = 0
        
        avg_result[element["name"]] = average
    
    return avg_result   # The result is a dict = {Restaurant name, Average Score}

        
# 2. Compute the minimum score for each restaurant.
def min_score(data_path):
    min_result = {}
    data = data_path
    i_score = 1000
    
    for element in data_path:
        for score in element["grades"]:   
            if score["score"] != None:
                if score["score"] < i_score:
                    i_score = score["score"]
        
        min_result[element["name"]] = i_score
               
    return min_result  # The result is a dict = {Restaurant name, Minimum Score}
    
                               
# 3. Compute the maximum score for each restaurant.
def max_score(data_path):
    max_result = {}
    data = data_path
    i_score = 0
    
    for element in data_path:
        for score in element["grades"]:   
            if score["score"] != None:
                if score["score"] > i_score:
                    i_score = score["score"]
        
        max_result[element["name"]] = i_score  

    return max_result   # The result is a dict = {Restaurant name, Maximum Score}
        

# 4. Compute the average score for each type of cuisine in each borough.
def create_boroughdict(data_path):   # We create a dictionary with all the boroughs, for a later use of it
    borough_dict = {}

    for element in data_path:
        if element["borough"] not in borough_dict:
            borough_dict[element["borough"]] = {}
    
    return borough_dict   



def avg_score_cuisine_borough(data_path):
    borough_avg = create_boroughdict(data_path)   # Import dict with boroughs
    cuisine_lst = []                  
    avg_lst = []
    
    for element in data_path:
        borough_avg[element['borough']][element['cuisine']] = 0   # Adds cuisine types to each borough of the dict

    for element in data_path:              # Applies avg_score
        for score in element["grades"]:
            if score["score"] != None:
                avg_lst.append(score["score"])

        if len(avg_lst) != 0:          
            average = sum(avg_lst)/len(avg_lst)
        else:                           
            average = 0
        
        borough_avg[element["borough"]][element["cuisine"]] = round((borough_avg[element["borough"]][element["cuisine"]] + average), 2)

    return borough_avg
        
                     
# 5. Compute the minimum score for each type of cuisine in each borough.
def min_score_cuisine_borough(data_path):
    borough_min = create_boroughdict(data_path)   # Import dict with boroughs               
    
    for element in data_path:
        borough_min[element['borough']][element['cuisine']] = 1000   # Adds cuisine types to each borough of the dict
    
    for element in data_path:
        i_score = 1000
        for score in element["grades"]:   
            if score["score"] != None:
                if score["score"] < i_score:
                    i_score = score["score"]
        
        if borough_min[element["borough"]][element["cuisine"]] > i_score:
            borough_min[element["borough"]][element["cuisine"]] = i_score
    
    return borough_min
        
        
# 6. Compute the maximum score for each type of cuisine in each borough.
def max_score_cuisine_borough(data_path):
    borough_max = create_boroughdict(data_path)   # Import dict with boroughs               
     
    for element in data_path:
        borough_max[element['borough']][element['cuisine']] = 0   # Adds cuisine types to each borough of the dict
    
    for element in data_path:
        i_score = 0
        for score in element["grades"]:   
            if score["score"] != None:
                if score["score"] > i_score:
                    i_score = score["score"]
        
        if borough_max[element["borough"]][element["cuisine"]] < i_score:
            borough_max[element["borough"]][element["cuisine"]] = i_score
            
    return borough_max


input_1 = input("Which information would you like to check? [1 to 6 in same order as in the title]")
input_1 = int(input_1)

if input_1 == 1:
    print("Showing Average Score for all Restaurants \n")
    print(avg_score(restaurant))

elif input_1 == 2:
    print("Showing Minimum Score for all Restaurants \n")
    print(min_score(restaurant))
    
elif input_1 == 3:
    print("Showing Maximum Score for all Restaurants \n")    
    print(max_score(restaurant))

elif input_1 == 4:
    print("Showing Average Score for for each type of cuisine in each borough \n")
    print(avg_score_cuisine_borough(restaurant))

elif input_1 == 5:
    print("Showing Minimum Score for for each type of cuisine in each borough \n")
    print(min_score_cuisine_borough(restaurant))

elif input_1 == 6:
    print("Showing Maximum Score for for each type of cuisine in each borough \n")
    print(max_score_cuisine_borough(restaurant))
else:
    print("Please insert a valid number.")
```
